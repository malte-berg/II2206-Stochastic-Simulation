```python
import numpy as np
import math
seed = 333
```

## 4.15

```python
np.random.seed(seed)

n = 1_000_000

c = 0

for i in range(n):

    u1 = np.random.uniform()
    u2 = np.random.uniform()

    even = ((math.floor(5 * u2) + 3) * 2)

    if u1 < 0.55:
        c += even - 1
    else: 
        c += even

res = c / n

print(res)
```

## 5.10

```python
np.random.seed(seed)

n= 100_000
c = 0

holders = 1000

for i in range(n):
    val = 0
    for j in range(holders):
        u1 = np.random.uniform()
        if u1 < 0.05:
            u2 = np.random.uniform()
            exp_val = -800 * math.log(u2)
            val += exp_val
    if val > 50_000:
        c += 1

res = c/n

print(res)
```

## 5.29

```python
np.random.seed(seed)

n = 1_000_000

fans = 0
rate = 5

for i in range(n):
    val = 0
    while val <= 1:
        ui = np.random.uniform()
        x = -(1/rate) * math.log(ui)
        val += x
        ub = np.random.uniform()
        fans += math.floor(ub * 21) + 20

res = fans / n

print(res)
```
