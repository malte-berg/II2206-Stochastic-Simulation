## 8.8

```python
# parts from HW1
def sum_num(n):
    np.random.seed(seed)
    res = 0
    squared_sum = 0

    for i in range(n):
        count, sum = 0, 0
        while sum <= 1:
            sum += np.random.uniform()
            count += 1
        res += count
        squared_sum += count ** 2
    return res, squared_sum

n = 1000
final_sum, squared_sum = sum_num(n)

s2 = (squared_sum - n * ((partial_sum/n) ** 2)) / (n - 1)
s = math.sqrt(s2)

var = s2/n

mean = final_sum / n

int_up = mean + 1.96 * (s / math.sqrt(n))
int_low = mean - 1.96 * (s / math.sqrt(n))

print(f"Sum: {final_sum}")
print(f"Est e (sample mean): {mean}")
print(f"Estimator var: {var}")
print(f"Interval, lower: {int_low}, up: {int_up} (alpha = 0.05, conf. int. 95%)")
```

```python
r = 100

bootstrap_list = [56, 101, 78, 67, 93, 87, 64, 72, 80, 69]
n = 10
a = -5
b = 5

mean = sum(bootstrap_list)/len(bootstrap_list)

def bootstrap(r):
    np.random.seed(seed)
    hit = 0

    for i in range(r):
        new_list = []
        for j in range(n):
            u_index = math.floor(n * np.random.uniform()) # not -1, 0 to n-1
            new_list.append(bootstrap_list[u_index])
        new_mean = sum(new_list)/len(new_list)
        if a < new_mean - mean < b:
            hit += 1

    return hit/r

print(bootstrap(r))
print(bootstrap(1000))
print(bootstrap(10000))
print(bootstrap(100000))
```
