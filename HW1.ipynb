## 3.3: $\int_0^1 \exp{\{e^x\}} \,dx$

```python
np.random.seed(seed)

n = 1_000_000

res = 0

for i in range(n):
    x = np.random.uniform()
    res += math.exp(math.exp(x))

z = res / n

print(z)
```

## 3.12: $N = \text{Minimum}\left\{ n: \sum_{i=1}^n U_i > 1 \right\}$

### i. Function

```python
def sum_num(n):
    np.random.seed(seed)
    res = 0

    for i in range(n):
        count, sum = 0, 0
        while sum <= 1:
            sum += np.random.uniform()
            count += 1
        res += count

    return res/n
```

### a. $E[N]$, $N$: 100 values

```python
print(sum_num(100))
```

### b. $E[N]$, $N$: 1000 values

```python
print(sum_num(1000))
```

### c. $E[N]$, $N$: 10000 values

```python
print(sum_num(10000))
```

### d. Numerically

```python
print(sum_num(10_000_000))
```

## 4.1: Give an efficient algorithm to simulate the value of a random variable X such that $P\{X = 1\} = 0.3, P\{X=2\} = 0.2, P\{X=3\}=0.35, P\{X=4\}=0.15$

```python
np.random.seed(seed)

u = np.random.uniform()

p = {1: 0.3, 2: 0.2, 3: 0.35, 4: 0.15}

X = 0

# efficient
if u < p[3]:
    X = 3
elif u < p[3] + p[1]:
    X = 1
elif u < p[3] + p[1] + p[2]:
    X = 2
else:
    X = 4

# inefficient

if u < p[1]:
    X = 1
elif u < p[1] + p[2]:
    X = 2
elif u < p[1] + p[2] + p[3]:
    X = 3
else:
    X = 4
```

## 4.7: Dice rolls

```python
np.random.seed(seed)

n = 1_000_000
sides = 6

throws = 0

for i in range(n):
    t = 0
    combos_rolled = []
    while len(combos_rolled) < 11:
        t += 1
        u1, u2 = np.random.uniform(), np.random.uniform()
        d1 = math.floor(u1 * sides) + 1
        d2 = math.floor(u2 * sides) + 1
        d_combo = d1 + d2
        if d_combo not in combos_rolled:
            combos_rolled.append(d_combo)
    throws += t

expected = throws / n

print(expected)
```
